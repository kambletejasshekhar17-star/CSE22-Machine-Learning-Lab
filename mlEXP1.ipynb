```python
#Task 12: Calculate maximum and average Fare using np.max() and np.mean().
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("train.csv")

# Convert Fare column to NumPy array
fare_array = df["Fare"].to_numpy()

# Calculate maximum and average Fare
print("Maximum Fare:", np.max(fare_array))
print("Average Fare:", np.mean(fare_array))
```

```python
#Task 13: Plot histogram of Age.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("train.csv")

# Fill missing Age values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Plot histogram
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=20, color="skyblue", edgecolor="black")

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Histogram of Age")

plt.show()
```

```python
#Task 14: Plot bar chart of Survival Count.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("train.csv")

# Count survival values
survival_count = df["Survived"].value_counts()

# Plot bar chart
plt.figure(figsize=(6,4))

sns.barplot(x=survival_count.index, y=survival_count.values, palette="viridis")

plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.title("Survival Count")

plt.show()
```

```python
#Task 15: Plot scatter graph of Age vs Fare.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("train.csv")

# Fill missing Age values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Create scatter plot
plt.figure(figsize=(8,5))

plt.scatter(df["Age"], df["Fare"], color="blue", alpha=0.5)

plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare Scatter Plot")

plt.show()
```
